```python
#Um caixa eletrônico possui notas de:

#R$ 100
#R$ 50
#R$ 20
#R$ 10
#R$ 5
#R$ 2
#R$ 1

#Solicite ao usuário o valor do saque e informe a menor
#quantidade possível de notas para realizar o pagamento.

#Exemplo:

#Digite o valor do saque: 287

#Saída esperada:

#2 nota(s) de R$100
#1 nota(s) de R$50
#1 nota(s) de R$20
#1 nota(s) de R$10
#1 nota(s) de R$5
#1 nota(s) de R$2

#Requisitos:
#- Utilizar laços de repetição.
#- Não utilizar listas.
#- Exibir apenas as notas utilizadas.

valor = int(input("Digite o valor do saque: "))

while valor > 0:
    if valor >= 100:
        nota = 100
    elif valor >= 50:
        nota = 50
    elif valor >= 20:
        nota = 20
    elif valor >= 10:
        nota = 10
    elif valor >= 5:
        nota = 5
    elif valor >= 2:
        nota = 2
    else:
        nota = 1

    quantidade = valor // nota
    print(f"{quantidade} nota(s) de R${nota}")

    valor = valor % nota
```
