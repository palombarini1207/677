```python
#Contagem simples, o usuário digita um número e o programa imprimi do zero até o número.

numero = int (input("digite um numero"))

for i in range (numero + i):
  print (i)
```

```python
# 2 - Contagem regressiva, o usuário um número e o programa imprimi do número até o zero.

numero = int(input("digite um numero: "))

for i in range(numero + i):
  print(numero - i)
```

```python
# 3 - Números pares, o usuário digita um número e o programa imprimi os números pares.

numero = int(input("digite um numero: "))

for i in range(numero + 1):
  if i % 2==0:
    print (i)
```

```python
# 4 - Números impares, o usuário digita um número e o programa imprimi os números ímpares.

numero = int(input("digite um numero: "))

for i in range(numero + 1):
  if i % 2==1:
    print (i)

```

```python
# 5 - Soma, o usuário digita um número e o programa soma os números do zero até o número.

numero = int(input("digite um numero: "))

soma = 0
for i in range(numero + 1):
  soma += 1
print(soma)
```

```python
# 6 - Tabuada, o usuário digita um número e o programa imprimi a tabuada do número.

numero = int (input("digite um numero: "))

if 1 <= numero <=10:
  print(f"tabuada do {numero}:")
  for i in range(1,11):
    print( f"{numero} x {i} = {numero * i}")
```

```python
# 7 - Contar letras, o usuário digita uma palavra e o programa conta as letras da palavra.

palavra = input("Digite uma palavra: ")
quantidade = len(palavra)

print(f"A palavra '{palavra}' tem {quantidade} letras.")
```

```python
# 8 - Contar vogais, o usuário digita uma palavra e o programa conta as vogais.

palavra = input("Digite uma palavra: ").lower()
vogais = "aeiou"
contador = 0

for letra in palavra:
    if letra in vogais:
        contador += 1

print (f"A palavra tem {contador} vogais.")
```

```python
# 9 - Maior número digitado, o usuário digita quatro números e o programa imprimi o maior número digitado.

numeros = []

for i in range(4):
    num = float(input(f"Digite o {i+1}º número: "))
    numeros.append(num)

maior_numero = max(numeros)

print(f"O maior número digitado foi: {maior_numero}")
```

```python
# 10 - O usuário digita um número entre 01 e 100 e o programa mostra quais são divisíveis por 3 e 5

numero = int(input("Digite um número entre 1 e 100: "))

print(f"Números entre 1 e {numero} divisíveis por 3 e 5:")

for i in range(1, numero + 1):
    if i % 3 == 0 and i % 5 == 0:
        print(i)
```
