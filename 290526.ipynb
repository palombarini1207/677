```python
# Escrever um jogo de advinhação. O usuário terá 03 chances para adivinhar um número 0 a 10.
# O programa dar uma mensagem se o palpite foi maior ou menor
# O programa avisar quente ou frio



import random

# Sorteia um número de 0 a 10
numero_secreto = random.randint(0, 10)

tentativas = 3

print("=== JOGO DE ADIVINHAÇÃO ===")
print("Tente adivinhar o número de 0 a 10.")
print("Você tem 3 chances!")

for tentativa in range(1, tentativas + 1):
    palpite = int(input(f"{tentativa}ª tentativa: "))

    # Acertou
    if palpite == numero_secreto:
        print(" Parabéns! Você acertou o número!")
        break

    # Quente ou frio
    diferenca = abs(palpite - numero_secreto)

    if diferenca <= 2:
        print(" Quente! Você está perto.")
    else:
        print(" Frio! Você está longe.")

    # Maior ou menor
    if palpite > numero_secreto:
        print("Seu palpite foi MAIOR que o número secreto.")
    else:
        print("Seu palpite foi MENOR que o número secreto.")

# Caso perca
if palpite != numero_secreto:
    print(f"7 Você perdeu! O número era {numero_secreto}.")
```
