```python
def verificar_numero(numero):
    print("Esse número é", numero)
num= int(input("Digite um número: "))
if num == 0:
   verificar_numero("Zero")
elif num < 0:
   verificar_numero("Negativo")
elif num > 0:
   verificar_numero("Positivo")
```
